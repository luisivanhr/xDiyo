```python
import re
```

```python
def _first_year(season_str: str) -> int:
    m = re.search(r'(\d{4})', str(season_str))
    if m:
        return int(m.group(1))
    m2 = re.search(r'(\d+)', str(season_str))
    if m2:
        return int(m2.group(1))
    raise ValueError(f"Cannot parse season year from: {season_str!r}")
```

```python
tests = [
    "2019/2020", "1999-00", "S2", "Season 03", "N/A", "20/21"
]
for t in tests:
    try:
        print(t, "->", _first_year(t))
    except Exception as e:
        print(t, "->", e)
```
