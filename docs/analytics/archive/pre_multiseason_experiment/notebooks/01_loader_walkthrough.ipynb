# Load a football season

Choose a season, look at its available tables, then load the ones your experiment needs.
[Loading guide](../docs/analytics/season_loading.md) · [Table reference](../docs/analytics/season_inspection.md)

```python
from pathlib import Path
from IPython.display import display
from xdiyo_analytics.data import inspect_season, load_season

project_root = Path('C:/Users/luisi/Documents/Programming/Python/xDiyo')
data_root = project_root / 'data/xDiyo_data'
season_name = 'Premier_League_24_25'
```

```python
catalogue = inspect_season(data_root, season_name)
catalogue[['rows', 'column_count', 'description']]
```

```python
print('Statistics columns:', catalogue.loc['statistics', 'columns'])
print('Pregame columns and types:', catalogue.loc['pregame', 'column_summary'])
```

## Load the tables you need

Tables stay separate. Choose a single name or a list of names; load `shots` separately when needed.
`pregame.position` is the provider's match-specific team position, not a complete league standings table.

```python
season = load_season(
    data_root, season_name, tables=['matches', 'statistics', 'pregame'],
)
{name: frame.shape for name, frame in season.tables.items()}
```

```python
display(season.matches[['event_id', 'home_id', 'away_id']].head(3))
display(season.statistics[['event_id', 'period', 'group_name', 'key', 'side', 'value']].head(4))
display(season['pregame'][['event_id', 'side', 'position']].head(3))
```
